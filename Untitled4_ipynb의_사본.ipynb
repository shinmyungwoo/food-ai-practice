```python
# 데이터 불러오기용 라이브러리
import pandas as pd
import numpy as np
from sklearn.datasets import fetch_openml

# 쌀(Rice) 데이터 불러오기 (data_id를 사용할 때는 version을 제외해야 합니다)
rice = fetch_openml(data_id=40995, as_frame=True)
df = rice.frame

print("데이터 크기:", df.shape)   # (몇 줄, 몇 칸)인지
df.head()   # 위에서 5줄 미리보기
```

```python
# label(정답) = 맞춰야 할 품종 (Cammeo / Osmancik)
# feature(힌트) = 나머지 측정값들 (쌀알 크기, 넓이 등)

# 보통 마지막 컬럼이 정답(label)이에요. 이름 확인부터!
print("전체 컬럼:", df.columns.tolist())
print("정답(label) 후보 = 맨 마지막 컬럼:", df.columns[-1])
print()

# 정답 컬럼 이름 (보통 'Class')
label_col = df.columns[-1]

X = df.drop(columns=[label_col])   # X = 힌트(feature)들만
y = df[label_col]                  # y = 정답(label)

print("힌트(X) 모양:", X.shape)     # (줄 수, 힌트 개수)
print("정답(y) 종류:", y.unique())  # 어떤 정답들이 있나
print()
print("정답 개수 세보기:")
print(y.value_counts())            # 각 품종이 몇 개씩 있나
```

```python
from sklearn.model_selection import train_test_split

# 정답이 글자('0','1')일 수 있으니 숫자로 통일
y = y.astype(int)

# train(공부용 80%) / test(시험용 20%) 나누기
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,        # 20%는 시험용
    random_state=42,      # 항상 같게 나뉘도록 고정 (★변경 금지★)
    stratify=y            # 불균형 비율을 train/test에 똑같이 유지
)

print("공부용(train):", X_train.shape)
print("시험용(test):", X_test.shape)
print()
print("train 정답 비율:")
print(y_train.value_counts(normalize=True))   # 비율로 확인
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score, classification_report

baseline = LogisticRegression(max_iter=1000, random_state=42)
baseline.fit(X_train, y_train)
base_pred = baseline.predict(X_test)

base_acc = accuracy_score(y_test, base_pred)
base_f1  = f1_score(y_test, base_pred, average="macro")

print("=== Baseline 성능 ===")
print("정확도(accuracy):", round(base_acc, 4))
print("macro F1:", round(base_f1, 4))
print()
print(classification_report(y_test, base_pred))
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score, classification_report

# 개선 모델 = 스케일링 + 가중치 조정
improved = make_pipeline(
    StandardScaler(),                          # 무기1: 데이터 크기 맞추기
    LogisticRegression(
        max_iter=1000,
        random_state=42,
        class_weight="balanced"                # 무기2: 소수 클래스에 힘 실어주기 ⭐
    )
)

improved.fit(X_train, y_train)
imp_pred = improved.predict(X_test)

imp_acc = accuracy_score(y_test, imp_pred)
imp_f1  = f1_score(y_test, imp_pred, average="macro")

print("=== 개선 모델 성능 ===")
print("정확도(accuracy):", round(imp_acc, 4))
print("macro F1:", round(imp_f1, 4))
print()
print(classification_report(y_test, imp_pred))
```
